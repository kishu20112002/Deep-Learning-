## 4 Create the model
- First step is we need to create the model by using Sequential class
- Once model created then we need to add layers to the model

```python
## Model creation 


# step 1 importing librares
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# step 2 loading Dataset
df=loadtxt("/Users/apple/Downloads/diabetes.csv",
                delimiter=',',
                skiprows=1)

# step 3 data preparetions 
X=df[:,0:8]
y=df[:,8]

# step 4 splitting the dataset is optional

# step 5 model creation 

model=Sequential()

# step 5 a  creation layers 

layer1=Dense(12,input_shape=(8,),activation="relu")
layer2=Dense(8,activation="relu")
layer3=Dense(1,activation="sigmoid")

# step 5 b  adding  layers to models 

model.add(layer1)
model.add(layer2)
model.add(layer3)

```

### 4.1 Input shape and activation function
- Fully connectes layers are defined using the Dense class
- The model expects rows of data with 8 features (input_shape =(8,)arguments
- The first hidden layer has 12 nodes and uses the relu actication function
- The second hidden layer has 8 nodes and uses the relu activation function
- The output layer has 1 node and using the sigmoid activation function
- The line of code , Dense layer is doing two things, creatinf first hidden and input layers

## 5 Compile the keras model
- Once the model created then we need to compile the model

```python
# Compile the keras model

# step 1 importing librares
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# step 2 loading Dataset
df=loadtxt("/Users/apple/Downloads/diabetes.csv",
                delimiter=',',
                skiprows=1)

# step 3 data preparetions 
X=df[:,0:8]
y=df[:,8]

# step 4 splitting the dataset is optional

# step 5 model creation 

model=Sequential()

# step 5 a  creation layers 

layer1=Dense(12,input_shape=(8,),activation="relu")
layer2=Dense(8,activation="relu")
layer3=Dense(1,activation="sigmoid")

# step 5 b  adding  layers to models 

model.add(layer1)
model.add(layer2)
model.add(layer3)

# step 5 model compiletion 

model.compile(loss="binary_crosseentropy"
              ,optimizer="adam"
              ,metrics=["accuracy"])

print("Model compiled")
```

### 5.1 Loss function optimiser and metrics 
- we need to provide loss function to evaluate a set of weight
- The optimizer is used to search through different weights for the network
- This loss is for a binary classification problem and is defined in Keras as "binary_crossentropy"
- we have given optimizer value as adam.
  - This is more efficient gradient descent algorithm
 
## 6 Fit the model
- Once model compiled then we need to train the model
- By using fit(..)method we can train the model

```python
# training the model

# step 1 importing librares
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


# step 2 loading Dataset
df=loadtxt("/Users/apple/Downloads/diabetes.csv",delimiter=',',skiprows=1)


#step 3 data preparetions
X=df[:,0:8]
y=df[:,8]

# step 4 splitting the dataset is optional

# step 5 model creation 

model=Sequential()

# step 5a model layer

layer1=Dense(12,input_shape=(8,),activation='relu')
layer2=Dense(8,activation="relu")
layer3=Dense(1,activation="sigmoid")

# step 5b add the model  

model.add(layer1)
model.add(layer2)
model.add(layer3)

# step 6 compiler

model.compile(optimizer="adam",
             loss="binary_crossentropy",
             metrics=["accuracy"])

# step 7 model traning

model.fit(X,y,epochs=10,batch_size=10)

print("Model trained")
```
